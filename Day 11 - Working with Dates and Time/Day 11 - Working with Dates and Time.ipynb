```python
import datetime

# A Date Calculator

def get_datetime_input():
    """Get and parse date and time input from the user."""
    while True:
        try:
            date_string = input("Enter a date string in this format '25 December, 2022': ").strip()
            time_string = input("Enter a time string in this format '14:30': ").strip()
            datetime_string = f"{date_string} {time_string}"
            return datetime.datetime.strptime(datetime_string, "%d %B, %Y %H:%M")
        except ValueError:
            print("Invalid date or time format. Please try again.")


def get_duration_input():
    """Get duration input from the user."""
    while True:
        try:
            days = int(input('Enter the number of days you want to add or subtract: '))
            hours = int(input('Enter the number of hours you want to add or subtract: '))
            minutes = int(input('Enter the number of minutes you want to add or subtract: '))
            return datetime.timedelta(days=days, hours=hours, minutes=minutes)
        except ValueError:
            print("Invalid input. Please enter integers for days, hours, and minutes.")


def get_operation():
    """Get the operation choice (Add or Subtract) from the user."""
    while True:
        operation = input("Do you want to add or subtract the duration? [Enter A / S]: ").strip().upper()
        if operation in ['A', 'S']:
            return operation
        print("Invalid choice. Please enter 'A' to Add or 'S' to Subtract.")


def main():
    # Get and parse datetime input
    parsed_datetime = get_datetime_input()
    
    # Get duration input
    duration = get_duration_input()
    
    # Get user operation
    operation = get_operation()
    
    # Perform calculation
    if operation == 'A':
        calculated_datetime = parsed_datetime + duration
    else:
        calculated_datetime = parsed_datetime - duration
    
    #Display results
    print("\n🗓️  **Date Calculation Result**")
    print(f"Initial Date and Time: {parsed_datetime.strftime('%Y-%m-%d %H:%M:%S')}")
    print(f"Calculated Date and Time: {calculated_datetime.strftime('%Y-%m-%d %H:%M:%S')}")


if __name__ == '__main__':
    main()
```

```python
'''
Calculate Age from Birthdate:

Develop a program that asks the user for their birthdate in the format "YYYY-MM-DD", calculates their age in years, and displays it.

Example Output:

Enter your birthdate (YYYY-MM-DD): 1990-06-15
You are 34 years old.
'''
# Define a function that gets date input from the user and parse it.
def get_date_string():
    """Get and parse date input from the user."""
    while True:
        try:
            # Prompt user to enter a date string in 'YYYY-MM-DD' format
            date_string = input("Enter a date string in this format 'YYYY-MM-DD': ").strip()
            # Convert the date string to a datetime object
            return datetime.datetime.strptime(date_string, "%Y-%m-%d")
        except ValueError:
            # Handle invalid date format
            print("Invalid date or time format. Please try again.")

# Get the current date
current_date = datetime.datetime.now()

# Get user's birthdate
birth_day = get_date_string()

# Calculate the age
age = current_date.year - birth_day.year - ((current_date.month, current_date.day) < (birth_day.month, birth_day.day))

# Display the result
print(f'\nYou are {age} years old.' )
```

```python
'''
Convert datetime to Timestamp and Vice Versa:
Create a script that converts the current datetime to a Unix timestamp and then converts that timestamp back to a datetime object, displaying both.

Example Output:
Current datetime: 2025-01-05 16:23:01
Unix Timestamp: 1736101381
Converted datetime: 2025-01-05 16:23:01
'''

current_datetime = datetime.datetime.now()

# Convert datetime to Unix timestamp
timestamp = datetime.datetime.timestamp(current_datetime)

# Convert Unix timestamp to datetime object
datetime_obj = datetime.datetime.fromtimestamp(timestamp)

# Display output
print(f"Current datetime: {current_datetime.strftime('%Y-%m-%d %H:%M:%S')}\nUnix Timestamp: {timestamp:2f}\nConverted datetime: {datetime_obj.strftime('%Y-%m-%d %H:%M:%S')}")
```

```python
'''
Format datetime Object into a Readable String:
Given a datetime object, format it into a string that reads "Sunday, January 5, 2025 at 04:23 PM".

Example Output:
Formatted datetime string: Sunday, January 5, 2025 at 04:23 PM
'''

def get_datetime_input():
    """Get and parse date and time input from the user."""
    while True:
        try:
            date_string = input("Enter a date string in this format '25 December, 2022': ").strip()
            time_string = input("Enter a time string in this format '14:30': ").strip()
            datetime_string = f"{date_string} {time_string}"
            return datetime.datetime.strptime(datetime_string, "%d %B, %Y %H:%M")
        except ValueError:
            print("Invalid date or time format. Please try again.")

# Format datetime object into string
date_object = datetime.datetime.now()

formatted_datetime_string = f"{date_object.strftime('%A, %B %d, %Y')} at {date_object.strftime('%I:%M %p')}"

# Display an output
print(f"Formatted datetime string: {formatted_datetime_string}")
```

```python
'''
Calculate Difference Between Two Dates:

Write a program that takes two dates as input and calculates the number of days between them.

Example Output:
Enter the first date (YYYY-MM-DD): 2025-01-01
Enter the second date (YYYY-MM-DD): 2025-01-05
Difference: 4 days
'''
# Define a function to get and parse a date input from the user
def get_parse_date_input(prompt):
    """Get and parse a date string input from the user."""
    while True:
        # Prompt the user for a date and parse it into a datetime object
        try:
            date = input(prompt)
            return datetime.datetime.strptime(date, "%Y-%m-%d")
        except ValueError:
            # Handle invalid date format
            print("Invalid input. Please enter date in the format (YYYY-MM-DD) ")

# Get the first date from the user
first_date = get_parse_date_input('Enter the first date (YYYY-MM-DD): ')

# Get the second date from the user
second_date = get_parse_date_input('Enter the second date (YYYY-MM-DD): ')

# Calculate the absolute difference between the two dates
difference_of_dates = abs(first_date - second_date)

# Display the difference in days
print(f"Difference : {difference_of_dates.days} days")




```

```python
# Display message to the user to select a region to show the available timezones and their abbreviations.
print("Here are common timezones grouped by their regions. \nSelect a the number attached to the region you want to print the current date and time.\n")

# Initialize the index to 1 for numbering the options and an empty list to store timezones and their abbreviations.
index = 1
timezone_list = []

# Loop through the timezones dictionary to display regions and their timezones along with abbreviations.
for key,value in timezones.items():
    # Print the region name (e.g., 🌍 UTC and General Timezones, 🇺🇸 America Timezones)
    print(f'{key}')

    # Loop through each timezone and its abbreviation in the current region.
    for tz,abbrev in value:
         # Print the numbered timezone with its abbreviation.
        print(f'{index}. {tz} ({abbrev})')
        # Append the tuple (timezone, abbreviation) to the timezone_list.
        timezone_list.append((tz,abbrev))
        
        index += 1
    print('\n')

# Prompt the user to input a number corresponding to a timezone.
user_choice = int(input('Enter a number: ')) - 1

# Check if the user's choice is valid (within the bounds of the timezone_list).
if 0 <= user_choice < len(timezone_list):
    user_tz,abbreviation = timezone_list[user_choice]

    # Get the current time for the selected timezone using pytz.
    time_zone = pytz.timezone(user_tz)
    date_time = datetime.datetime.now(time_zone)
    print(f'Current time in {user_tz}: {date_time.strftime('%Y-%m-%d %I:%M:%S %p')} {abbreviation}')

# If the user entered an invalid choice (outside the range of available options), print an error message.
else:
    print("Invalid choice, please try again.")


```
