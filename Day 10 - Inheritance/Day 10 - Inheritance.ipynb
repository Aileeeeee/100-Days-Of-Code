```python
'''
🩺 Healthcare OOP Exercise: Hospital Management System
Objective:
Practice Object-Oriented Programming (OOP) concepts such as classes, objects, inheritance, polymorphism, and encapsulation by building a simplified hospital management system.

📝 Exercise Description
You are tasked with creating a basic Hospital Management System in Python. The system should allow managing Patients, Doctors, and Appointments using OOP principles.

⚙️ Requirements
Class: Person
Attributes: name, age, gender
Method: display_person_info()

Class: Patient (inherits from Person)
Additional Attributes: patient_id, disease
Method: get_patient_info() → Displays patient details including disease.

Class: Doctor (inherits from Person)
Additional Attributes: doctor_id, specialization
Method: get_doctor_info() → Displays doctor details including specialization.

Class: Appointment
Attributes: appointment_id, patient (Patient object), doctor (Doctor object), date
Method: schedule_appointment() → Displays appointment details.

🧠 Tasks
Create instances of Patient and Doctor.
Schedule an Appointment between a patient and a doctor.
Display all details of the appointment, including the patient’s disease and the doctor’s specialization.
'''
```

```python
class Person:
    """
    A class to represent a person."""

    def __init__(self,name,age = 0,gender= 'N/A'):

        """
        Initializes a Person instance.

        Args:
            name (str): The name of the person.
            age (int, optional): The age of the person. Defaults to 0.
            gender (str, optional): The gender of the person. Defaults to 'N/A'.
        """
        
        self.name = name
        self.age = age
        self.gender = gender

    def display_person_info(self):
        
        """
        Prints the personal details of the person.
        """
        
        print(f"Name: {self.name}\nAge: {self.age}\nGender: {self.gender}")

class Patient(Person):
   
    """
    A class that represents a Patient and inherits from the Person class.  
    """
    
    def __init__(self,name,age,gender,patient_id,disease):

        """
        Initializes a Patient instance.

        Args:
            name (str): The name of the patient.
            age (int): The age of the patient.
            gender (str): The gender of the patient.
            patient_id (str): Unique identifier for the patient.
            disease (str): The disease or condition the patient has.
        """
        super().__init__(name,age,gender)
        self.patient_id = patient_id
        self.disease = disease

    def get_patient_info(self):

        """
        Prints the detailed information of the patient.
        """
        print('***Patient details***')
        self.display_person_info()
        print(f"Patient ID: {self.patient_id}\nDisease: {self.disease}")
        
class Doctor(Person):
    """ 
    A class that represents a Doctor and inherits from the Person class.
    """
    
    def __init__(self,name,doctor_id,specialization):

        """
        Initializes a Doctor instance.

        Args:
            name (str): The name of the doctor.
            doctor_id (str): Unique identifier for the doctor.
            specialization (str): The area of specialization of the doctor.
        """
        super().__init__(name)
        self.doctor_id = doctor_id
        self.specialization = specialization

    def display_person_info(self):
        
        """
        Returns a formatted string with the doctor's name prefixed with 'Dr.'.
        """
        return f'Name: Dr. {self.name}'

    def get_doctor_info(self):
        
        """
        Prints the detailed information of the doctor.
        """
        print('Doctor details:')
        self.display_person_info()
        print(f"Doctor ID: {self.doctor_id}\nSpecialization: {self.specialization}")

class Appointment:

    """
    A class to represent an Appointment.
    """

    def __init__(self,appointment_id,patient,doctor,date):

        """
        Initializes an Appointment instance.

        Args:
            appointment_id (str): Unique identifier for the appointment.
            patient (Patient): The patient involved in the appointment.
            doctor (Doctor): The doctor involved in the appointment.
            date (str): The date of the appointment.
        """
        self.appointment_id = appointment_id
        self.patient = patient
        self.doctor = doctor
        self.date = date

    def appointment_details(self):
        """
        Prints the details of the appointment, including patient and doctor information.
        """
        
        print('\nAppointment Details:')
        print(f"Appointment ID: {self.appointment_id}\nPatient Name: {self.patient.name}\nPatient ID: {self.patient.patient_id}\nPatient Age: {self.patient.age}\nPatient Disease: {self.patient.disease}\nDoctor Name: Dr. {self.doctor.name}\nDoctor Specialization: {self.doctor.specialization}\nAppointment Date: {self.date}")
    
```

```python
# Import datetime
from datetime import datetime

# Global variable for generating sequential IDs
sequential_id = 100

# Helper function to get user's name 
def get_name():
    """
    Prompts the user to input their first and last name, then formats it.
    """
    first_name = input('Enter First name: ').strip().capitalize()
    last_name = input('Enter last name: ').strip().capitalize()
    return f'{first_name} {last_name}'

def generate_patient_id(name):
    """
    Generates a unique patient ID using the patient's initials and a global sequential ID.

    Args:
        name (str): The full name of the patient.
    """

    global sequential_id

    first_name,last_name = name.split()

    first_initial = first_name[0].upper()
    last_initial = last_name[0].upper()

    unique_id = f'PAT-{first_initial}{last_initial}{sequential_id}'

    # Increment sequential ID for uniqueness
    sequential_id += 1

    return unique_id

def generate_doctor_id(name):

    """
    Generates a unique doctor ID using the patient's initials and a global sequential ID.

    Args:
        name (str): The full name of the patient.
    """

    global sequential_id

    first_name,last_name = name.split()

    first_initial = first_name[0].upper()
    last_initial = last_name[0].upper()

    unique_id = f'DOC-{first_initial}{last_initial}{sequential_id}'

    sequential_id += 1

    return unique_id

def format_date(datetime_object):
    """
    Formats a datetime object into a human-readable string with an ordinal suffix for the day.

    Args:
        datetime_object (datetime): A datetime object.
    """

    # Extract the date part
    date = datetime_object.strftime('%d')

    # Converting to integer
    d = int(date)

    # Add the ordinal suffix
    date = date + ('th' if 4<= d % 100 <= 20 else {1 :'st', 2 : 'nd', 3 : 'rd'}.get(d % 10, 'th'))

    # Create full date and time
    full_date_time = datetime_object.strftime(f"%A, {date} %B, %Y  %I:%M %p")

    return full_date_time
    



def main():

    print ('🏥 **Welcome to Westend Hospital Hospital Management System** 🏥')

    # Get Patient details
    print("\nGetting patient's details.........")
    patient_name = get_name()
    patient_age = int(input('Enter Age: '))
    patient_gender = input('Enter Gender')
    patient_id = generate_patient_id(patient_name)
    disease = input("Enter Patient's Disease condition: ").strip().capitalize()

    # Get Doctor details
    print("\nGetting doctor's details.........")
    doctor_name = get_name()
    doctor_id = generate_doctor_id(doctor_name)
    specialization = input("Enter Doctor's Specialization: ").strip().capitalize()

    # Get appointment date 
    date = input("Enter appointment date in 'dd/mm/yyyy' format: ")
    time = input("Enter appointment time in 'Hour:Minute' format: ")

    # Convert date and time to a datetime object
    day,month,year = map(int,date.split('/'))
    hour,minute = map(int,time.split(':'))

    date_time = datetime(year,month,day,hour,minute)

    formatted_date = format_date(date_time)

    # Generate appointment ID
    appointment_id = f"APPT-{sequential_id}"
    
    # Create Patient, Doctor, and Appointment objects
    patient = Patient(patient_name,patient_age,patient_gender,patient_id,disease)
    doctor = Doctor(doctor_name,doctor_id,specialization)  

    appointment = Appointment(appointment_id,patient,doctor,formatted_date)

    # Display appointment details
    appointment.appointment_details()
    

# Run the program
if __name__ == '__main__':
    main()
```
