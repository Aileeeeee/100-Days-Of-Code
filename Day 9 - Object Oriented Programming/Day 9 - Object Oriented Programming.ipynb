```python
'''
Exercise 1: Custom Class: Create a class unrelated to vehicles, animals, or people. Include:
At least 3 attributes.
A constructor and a method to display object details.
'''

class Book:
    
    """
    A class to represent a book with attributes such as title, author, 
    number of pages, genre, and reading status.
    """

    
    def __init__(self,title,author,pages,genre,is_read = False):

        """
        Initialize a new book instance.
        
        Args:
            title (str): The title of the book.
            author (str): The author of the book.
            pages (int): The number of pages in the book.
            genre (str): The genre of the book.
            is_read (bool, optional): Whether the book has been read. Defaults to False.
        """
        
        self.title = title
        self.author = author
        self.pages = pages
        self.genre = genre
        self.is_read = is_read

    def read(self):

        """
        Mark the book as read and print a confirmation message.
        """
        
        self.is_read = True
        print(f'{self.title} has been marked as read!')

    def get_summary(self):
        
        """
        Display a summary of the book, including title, author, pages, genre, and read status.
        """
        print(f"\nBook : {self.title}\nAuthor : {self.author}\nPages : {self.pages}\nGenre : {self.genre}\nRead_Status : {self.is_read}")
        
        
atomic_habits = Book('Atomic Habits', 'James Clear', '319','Self Development',True)

atomic_habits.read()
atomic_habits.get_summary()

```

```python
'''
Exercise 2: Classy Vehicles: Create a vehicle class with attributes like type, year, color, doors. Add methods:
display_info() to show vehicle details.
has_wheels() to return True or False based on wheel count.
'''

class Vehicle:

        """
        Represents a vehicle with various attributes such as type, year, color, door count, and wheel count.
        Provides functionality to display vehicle information and check if it has wheels.
        """

    def __init__(self,vehicle_type,year,color,door_count,wheel_count,has_wheel):

        """
        Initializes a new Vehicle instance with type, year, color, door count, and wheel count.
        
        Args:
            vehicle_type (str): The type/category of the vehicle (e.g., Car, Truck, Motorcycle).
            year (int): The manufacturing year of the vehicle (not earlier than 1886).
            color (str): The color of the vehicle.
            door_count (int): The number of doors on the vehicle.
            wheel_count (int): The number of wheels on the vehicle.
        """

        # Validate vehicle_type argument
        if not isinstance(vehicle_type,str):
            print('Enter a non_empty valid string')
        self.vehicle_type = vehicle_type

        # Validate year argument 
        if not isinstance(year,int) or year < 1886:
            print('Enter a non_empty valid year not earlier than 1886') # Since the first automobile was created in 1886
        self.year = year

        # Validate color argument 
        if not isinstance(color,str) :
            print('Enter a non_empty valid string for the color ') 
        self.color = color

        # Validate door_count argument
        if not isinstance(door_count,int) or door_count < 0:
            print('Enter a non_negative numerical value') 
        self.door_count = door_count

        # Validate door_count argument
        if not isinstance(wheel_count,int) or wheel_count < 0:
            print('Enter a non_negative numerical value') 
        self.wheel_count = wheel_count
        

    def display_info(self):
        
        """
        Displays detailed information about the vehicle, including type, year, color, doors, and wheels.
        """

        # Applying conditional formatting to the door_count formatted text
        door_count_formatted_text = f"Has {self.door_count} door{'s' if self.door_count != 1 else ''}"
        
        print(f"Vehicle Type : {self.vehicle_type}\nYear : {self.year}\nColor : {self.color}\nDoor : {door_count_formatted_text}")

    def has_wheels(self):

        """
        Checks if the vehicle has wheels.
        
        Returns:
            bool: True if the vehicle has one or more wheels, False otherwise.
        """
        if self.wheel_count > 0:
            print(f'\nThis vehicle has {self.wheel_count} wheel{'s' if self.wheel_count != 1 else ''}.') 
            return True
        else:
            print(f'\nThis vehicle has no wheels.') 
            return False
            
        
```

```python
'''
Exercise 3 : 
Simple Calculator: Build a calculator with:
Operations as methods (add, subtract, multiply, divide).
A loop to take user inputs and handle "Quit" to exit.
'''

class Calculator:

    """
    A simple calculator class that supports basic arithmetic operations,
    maintains a history of calculations, and allows changing the model name.
    """

    model_name = 'Casio FX-991ES Plus' # Default calculator model

    def  __init__(self,mode = 'Standard'):
        
        """
        Initializes the calculator with a specific mode and an empty history.
        
        Args:
            mode (str): The mode of the calculator (e.g., 'Standard', 'Scientific').
        """
        
        self.mode = mode
        self.history = []

    def add(self,first_number,second_number):
        
        """
        Performs addition of two numbers and records the calculation in history.
        
        Args:
            first_number (float): The first number.
            second_number (float): The second number.
        
        Returns:
            float: The result of the addition.
        """
        result = first_number + second_number
        self.history.append(f'{first_number} + {second_number} = {result}')
        return result

    def subtract(self,first_number,second_number):
        
        """
        Performs subtraction of two numbers and records the calculation in history.
        
        Args:
            first_number (float): The first number.
            second_number (float): The second number.
        
        Returns:
            float: The result of the subtraction.
        """
        
        result = first_number - second_number
        self.history.append(f'{first_number} - {second_number} = {result}')
        return result

    def multiply(self,first_number,second_number):
        
        """
        Performs multiplication of two numbers and records the calculation in history.
        
        Args:
            first_number (float): The first number.
            second_number (float): The second number.
        
        Returns:
            float: The result of the multiplication.
        """
        
        result = first_number * second_number
        self.history.append(f'{first_number} * {second_number} = {result}')
        return result

    def divide(self,first_number,second_number):

        """
        Performs division of two numbers and records the calculation in history.
        
        Args:
            first_number (float): The numerator.
            second_number (float): The denominator.
        
        Returns:
            float: The result of the division.
        """
        result = first_number / second_number
        self.history.append(f'{first_number} / {second_number} = {result}')
        return result
    
    def show_history(self):

        """
        Displays the history of all calculations performed.
        """
         print('\n\nCalculation history:')
         for index,entry in enumerate (self.history,1):
            print(entry)
        

    @classmethod

    """
    Changes the model name of the calculator.
        
    Args:
    new_model_name (str): The new model name for the calculator.
    """
    
    def change_model(cls,new_model_name):
        cls.model_name = new_model_name



def main():
    calculator = Calculator(mode = 'Standard')
   
    user_action = ["Exit my 'Simple Calculator'" , "Perform an operation on my 'Simple Calculator'", "Check my 'Simple Calculator' history","Change my 'Simple Calculator' model"]
   
    mathematical_operations = ['Addition','Subtraction','Multiplication','Division','Return to main menu']
    
    print("Welcome to my 'Simple Calculator'")
    
    while True:
        # Display main menu options
        print('\n')
        for index,operation in enumerate(user_action,1):
            print(f'{index}. {operation}')
                
        user_input = int(input("\nWhat action do you want to carry out on my 'Simple Calculator'? "))

        if user_input == 1:
            print("\n\nGoodbye! Thanks for using my 'Simple Calculator'.")
            break
            
        elif user_input == 2:
            # Display mathematical operations
            print("You can perform any of the operations on my 'Simple Calculator'")
            for index,operation in enumerate(mathematical_operations,1):
                print(f'{index}. {operation}')

            while True:
               
                operation = input('\n\nWhat mathematical operation do you want to carry out? ').strip().lower()

                if operation == '5':
                    break

                first_number = int(input('Enter a number: '))
                second_number = int(input('Enter another number: '))
            
                
                if operation == '1':
                    print(f"Result: {calculator.add(first_number, second_number)}")
                    break

                elif operation == '2':
                    print(f"Result: {calculator.subtract(first_number, second_number)}")
                    break

                elif operation == '3':
                    print(f"Result: {calculator.multiply(first_number, second_number)}")
                    break

                elif operation == '4':
                    if second_number == 0:
                        print('Division by zero is not allowed!')
                    else:
                        print(f"Result: {calculator.divide(first_number, second_number)}")
                    continue
                else:
                    print(f'Invalid operation!')

        elif user_input == 3:
            # Show calculation history
            calculator.show_history()
           

        elif user_input == 4:
            # Change calculator model
            new_model = input('\n\nEnter the calculator model you want to use').strip()
            calculator.change_model(new_model)
            print(f'The calculator model has been changed to {Calculator.model_name}.')
  
        else:
            print('\n\nInvalid response! Enter a valid input')
            

if __name__ == "__main__" :
    main()
```

```python
'''
Exercise: Patient Monitoring System
Create a Patient class to manage information about patients in a hospital or clinic. The class should have the following attributes and methods:

Attributes:
- name (str): The name of the patient.
- age (int): The age of the patient.
- gender (str): The gender of the patient.
- medical_conditions (list): A list of medical conditions the patient has.
- is_admitted (bool): A flag to indicate if the patient is currently admitted.

Methods:
- display_info():
Prints the patient's details, including their name, age, gender, medical conditions, and admission status.

- add_condition(condition):
Adds a medical condition to the patient's list of conditions.

- discharge():
Updates the is_admitted attribute to False.

- admit():
Updates the is_admitted attribute to True.

- check_critical():
Checks if any of the conditions in medical_conditions match a predefined list of critical conditions (e.g., "Heart Attack", "Stroke") and prints an appropriate message.

Bonus Challenge
Create a Hospital class to manage multiple patients with the following:
- patients (list): A list of Patient objects.

Methods to:
- Add a new patient.
- Remove a patient by name.
- Display a list of all admitted patients.
'''
```

```python
# List of critical medical condition
critical_medical_conditions = [
    "Heart Attack",
    "Stroke",
    "Cancer",
    "Severe Asthma Attack",
    "Kidney Failure",
    "Liver Failure",
    "Sepsis",
    "Severe Anemia",
    "Pneumonia",
    "Acute Respiratory Distress Syndrome (ARDS)",
    "Multiple Organ Failure",
    "Severe Burns",
    "Major Trauma",
    "Diabetic Ketoacidosis",
    "Pulmonary Embolism",
    "Severe Hypertension Crisis",
    "Meningitis",
    "Encephalitis",
    "HIV/AIDS (Advanced Stage)",
    "Severe Allergic Reaction (Anaphylaxis)",
    "Acute Pancreatitis",
    "Intracranial Hemorrhage",
    "Coma",
    "Severe Dehydration",
    "Severe Malnutrition",
    "Seizures"
]


class Patient:

    """
    Represents a patient in a medical setting.
    """

    def __init__(self,name,age,gender,is_admitted):

        """
        Initializes a new Patient instance.

        Args:
            name (str): The name of the patient.
            age (int): The age of the patient.
            gender (str): The gender of the patient.
            is_admitted (bool): Indicates if the patient is currently admitted.
        """
        self.name = name
        self.age = age
        self.gender = gender
        self.is_admitted = is_admitted
        self.medical_conditions = []

    def display_info(self):

        """
        Displays detailed information about the patient.

        This method prints the patient's name, age, gender, admission status, 
        medical conditions, and critical condition check results.

        """
        admission_status = 'Admitted' if self.is_admitted else 'Not admitted'
        patient_conditions = ','.join(self.medical_conditions) if self.medical_conditions else "No conditions were found."
       
        print(f'Patient Information:\nName: {self.name}\nAge: {self.age}\nGender: {self.gender}\nAdmission Status: {admission_status}\nMedical Conditions: {patient_conditions}\n\n{self.check_critical()}')

    def add_condition(self,condition):
        
        """
        Adds a medical condition to the patient's record if it doesn't already exist.
        """
        
        if condition not in self.medical_conditions:
            self.medical_conditions.append(condition)
            print(f"The medical condition '{condition}' has been added sucessfully to patient's medical records.")
        else:
            print(f"The medical condition '{condition}' already exists in patient's medical records")       
     
    def remove_condition(self,condition):
        
        """
        Removes a medical condition from the patient's record if it exists.
        """
        
        if condition in self.medical_conditions:
            self.medical_conditions = [ailment for ailment in self.medical_conditions if ailment != condition]
            print(f"The medical condition '{condition}' has been removed sucessfully from patient's medical records.")
            return self.medical_conditions   
        else:
            print(f"The medication condition '{condition}' does not exist in patient's medical records.")
      
    def discharge(self):
       
        """
        Updates the patient's admission status to 'not admitted'.
        """
        # Set the patient's admission status to False
        self.is_admitted = False

    def admit(self):
        """
        Updates the patient's admission status to 'admitted'.

        This method is used to mark the patient as currently admitted to the medical facility.
        """
        # Set the patient's admission status to True
        self.is_admitted = True

    def check_critical(self):

        """
        Checks if the patient has any critical medical conditions.

        Critical conditions are determined by comparing the patient's medical conditions
        with a predefined list of critical conditions.
        """
        
        for condition in self.medical_conditions:
            if condition.lower() in [critical_condition.lower() for critical_condition in critical_medical_conditions]:
                return f"⚠️ Alert: Patient's condition is CRITICAL due to '{condition}'. Immediate attention required!"
        
        return f"✅ Patient's condition is stable. No critical conditions detected."

```

```python
class Hospital:
    
    """
    A class to represent a hospital and manage its patients.
    
    """

    def __init__(self,name):
      
        """
        Initializes the hospital with a name and an empty list of patients.
        """
        
        self.name = name
        
        # Initialize an empty list to hold admitted patients
        self.patients = []

    def admit_patient(self,patient):
        
        """
        Admits a patient to the hospital.
        """
        # Check if the patient is already admitted
        if patient not in self.patients:
            # Add the patient to the hospital's list of patients
            self.patients.append(patient)
            print(f"Patient '{patient.name}' has been sucessfully admitted in {self.name}.")
        else:
            print(f"Patient '{patient.name}' is already admitted to {self.name} Hospital.")

    def discharge_patient(self,name):
       
        """
        Remove a patient from the hospital system by name.
        """
        # Iterate through the list of admitted patients to find a match by name
        for patient in self.patients:
            if patient.name == name:
                 # Remove the patient from the hospital's list
                 self.patients.remove(patient)
                 print(f"✅ **Patient '{name}' has been successfully removed from the hospital system.**")
                 # Exit the method after successfully discharging the patient
                 return
        # Inform the user if the patient was not found in the hospital system
        print(f"⚠️ **Patient '{name}' not found in the hospital system.**")
                
                  
    def display_patients(self):
        
        """
        Displays a list of all patients currently admitted in the hospital along with their details.

        For each patient, the following information is shown:
        - Name
        - Age
        - Gender
        - Admission Status (Admitted or Not Admitted)
        - Medical Conditions (if any)
        - Critical condition status based on predefined critical conditions.
        """
        # Check if there are no patients in the hospital
        if not self.patients:
            print("⚠️ No patients currently admitted in the hospital.")
            return

        # Display the header for the patient list
        print("\n🏥 **List of Patients:r**\n")
        
        # Iterate through each patient in the list of admitted patients
        for patient in self.patients:
            admission_status = 'Admitted' if patient.is_admitted == True else 'Not admitted'
            patient_conditions = ','.join(patient.medical_conditions) if patient.medical_conditions else "No conditions were found."

            # Print the details of the current patient
            print(f'Name: {patient.name}\nAge: {patient.age}\nGender: {patient.gender}\nAdmission Status: {admission_status}\nMedical Conditions: {patient_conditions}\n\n{patient.check_critical()}')
            print('\n')
```

```python
# Helper function to get patient's full name
def get_patient_name():
    first_name = input("\nEnter patient's First name: ").strip().capitalize()
    last_name = input("Enter patient's last name: ").strip().capitalize()
    return f"{first_name} {last_name}"

# Main Program
def main():
    
    # Create a Hospital instance
    hospital = Hospital("Westend Hospital")

    # Display the  Hospital main menu
    while True:
        print("\n🏥 **Hospital Management System** 🏥")
        print("1. Add a Patient")
        print("2. Admit an existing patient")
        print("3. Search and display patient info")
        print("4. Discharge a patient")
        print("5. Add a Medical Condition to a Patient")
        print("6. Remove a Medical Condition from a Patient")
        print("7. Display All Patients")
        print("8. Exit")

        # Get user input for menu choice
        choice = input("\nEnter your choice: ").strip()

        # Add a new patient
        if choice == '1':
            # Retrieve the patient's name
            name = get_patient_name()
            age = int(input("Enter patient's age: "))
            gender = input("Enter patient's gender: ").strip().capitalize()
            is_admitted = input("Is the patient admitted? (yes/no): ").strip().lower() == 'yes'

            # Create a new Patient instance and admit to the hospital
            patient = Patient(name, age, gender, is_admitted)
            hospital.admit_patient(patient)

        # Admit an existing patient
        elif choice == '2':

            # Retrieve the patient's name
            name = get_patient_name()

            # Search for the patient in the hospital's records
            for patient in hospital.patients:
                # If patient found
                if patient.name == name:
                    
                    # Check if the patient is already admitted
                    if patient.is_admitted:
                        print(f"⚠️ Patient '{name}' is already admitted.")
                    # Admit the patient
                    else:
                        patient.admit()
                        print(f"✅ Patient '{name}' was found in the system and is now admitted.")
                    break
                    
            # If patient not found
            else:
                print(f"🆕 Patient '{name}' not found in the system.")
                create_new = input("Do you want to create a new record for this patient? (yes/no): ").strip().lower()

                # Create a new record if user agrees
                if create_new == 'yes':
                    age = int(input("Enter patient's age: "))
                    gender = input("Enter patient's gender: ").strip().capitalize()
                    patient = Patient(name, age, gender, is_admitted=True)
                    hospital.admit_patient(patient)

                # Do nothing if user declines
                elif create_new == 'no':
                    print("🚫 No new record was created. Returning to the main menu.")
                # Handle invalid input
                else:
                    print("⚠️ Invalid choice. Returning to the main menu.")

        # Search and display patient information
        elif choice == '3':
            # Get the patient's name
            name = get_patient_name()
            # Search for the patient in the records
            for patient in hospital.patients:
                # If found, display patient details
                if patient.name == name:  
                    print(f"✅ Patient '{name}' found! Displaying details:\n")  
                    patient.display_info()  
                    break
                    
            # If not found, display an appropriate message
            else:  
                print(f"⚠️ Patient '{name}' not found in the system.")  

        # Discharge a patient
        elif choice == '4':
            name = get_patient_name()
            hospital.discharge_patient(name)

        # Add a medical condition to a patient's record
        elif choice == '5':
            name = get_patient_name()
            condition = input("Enter the medical condition to add: ").strip().capitalize()
            for patient in hospital.patients:
                if patient.name == name:
                    patient.add_condition(condition)
                    break
            else:
                print(f"⚠️ Patient '{name}' not found.")

        # Remove a medical condition from a patient's record
        elif choice == '6':
            name = get_patient_name()
            condition = input("Enter the medical condition to remove: ").strip().capitalize()
            for patient in hospital.patients:
                if patient.name == name:
                    patient.remove_condition(condition)
                    break
            else:
                print(f"⚠️ Patient '{name}' not found.")

        # Display all patients in the hospital
        elif choice == '7':
            hospital.display_patients()

        # Exit the system
        elif choice == '8':
            print("👋 Exiting the system. Goodbye!")
            break

        # Handle invalid menu choices
        else:
            print("⚠️ Invalid choice. Please try again.")
        
        # Pause for better UX
        input("\nPress Enter to continue...")

# Run the program
if __name__ == '__main__':
    main()
```
