```python

'''
Read and Print CSV Data
Task: Write a Python program that reads a CSV file and prints each row.
Goal: Understand how to read and print the contents of a CSV file.

'''

import csv

# Try to open and read the file

file_path = 'data.csv'
try:
    with open(file_path, mode='r') as file:
        reader = csv.reader(file)
        for row in reader:
            print(row)
except FileNotFoundError:
    print(f"Error: The file {file_path} was not found.")
except Exception as e:
    print(f"An error occurred: {e}")


```

```python
'''
 Merge CSV and JSON Data
Task: Write a Python program that reads data from both a CSV file and a JSON file, 
then merges the data and outputs it in a new CSV file. 
For example, you could merge employee data from a CSV and department data from a JSON file.

'''

import csv
import json

# Read the CSV file
employees = []
with open('employees.csv', mode='r') as file:
    reader = csv.DictReader(file)
    for row in reader:
        employees.append(row)

# Read the JSON file
with open('departments.json', 'r') as file:
    data = json.load(file)

# Merge the data based on the employee ID
for employee in employees:
    for department in data['employees']:
        if str(employee['id']) == str(department['id']):
            employee['department'] = department['department']

# Write the merged data to a new CSV file
with open('merged_data.csv', mode='w', newline='') as file:
    fieldnames = ['id', 'name', 'age', 'department']
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(employees)

print("Data merged and written to 'merged_data.csv'.")
```
