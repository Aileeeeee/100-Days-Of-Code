```python
'''
Exercise 1: Typing Numbers
Create a program that prompts the user to enter a number and then displays the type of the
number entered (e.g., complex, integer, or float). For example, if the user enters 6, the output
should be int.
Implement appropriate exception handling for situations where the user enters a string that
cannot be converted to a number.
'''

number = input('Enter a number (complex,integer or float: ')

try:
    if 'j' in number :
        num = complex(number)
        print(f"{number} is of the '{type(num).__name__}' value data type.")
    elif '.' in number :
        num = float(number)
        print(f"{number} is of the '{type(num).__name__}' value data type.")

    else:
        num = int(number)
        print(f"{number} is of the '{type(num).__name__}' value data type.")
except ValueError:
    print(f"The entered value '{number}' is a string and cannot be converted to a number.")

```

```python
'''
Exercise 2: Current Value
Write a program that calculates and displays the current value of a deposit for a given initial
deposit, interest rate, number of times interest is calculated per year, and the number of years
since the initial deposit.

The program should prompt the user for each of the values and use the following formula to
calculate the current value of the deposit:

V = P(1 + r/n)^nt
Where:
V: Value
P: Initial deposit
r: Interest rate as a fraction (e.g., 0.05)
n: The number of times per year interest is calculated
t: The number of years since the initial deposit

The program should display each of the values entered in a meaningful way so that the user
can easily see what each value represents, along with the results of the calculation.
Implement appropriate exception handling for each input where the user might enter a string
value that cannot be used as a number. In particular, consider the following:
The user enters a percentage value like “5%” instead of “0.05” for the interest rate.
The user enters “0” for n, resulting in a division by 0 error.
'''


def get_float_input(prompt):

    '''
    Prompts the user for a numeric value and handles invalid inputs.
    If the input contains a '%' symbol, it converts it to a decimal (e.g., '5%' -> 0.05).

    '''

    while True:
        
        # Get input from the user
        user_input = input(prompt)
        
        try:
            # Validate user input
            if '%' in user_input:
                user_input = user_input.strip('%')
                return float(user_input) / 100
            return float(user_input)

        # Handle invalid input (e.g., non-numeric values)
        except ValueError:
            print("Invalid input. Please enter a numeric value.")


def main():
    
    # Display welcome message to user
    print('Welcome to the Compound Interest Calculator')

    # Get user input
    P = get_float_input("Enter the initial deposit (P): ")
    r = get_float_input("Enter the annual interest rate as a fraction (e.g., 0.05 for 5%): ")
    t = get_float_input("Enter the number of years (t): ")

    # Prompt user for the number of times per year interest is calculated
    while True:
        n = get_float_input('Enter the number of times per year interest is calculated (n): ')

        # Prevent division by zero by ensuring n is not zero
        if n == 0:
            print("The value of 'n' cannot be zero . Please enter a valid number.")
        else:
            break

    # Calculate the current value of the deposit using the compound interest formula
    V = P * (1 + (r/n)) ** (n * t)


    # Display the results in a user-friendly format
    print('\n===== Calculation Results =====')
    print(f"Initial deposit (P): ${P:.2f}\nAnnual interest rate (r): {r:.2f}%\nCompounding Frequency(n): {int(n)} times per year\nNumber of years since the initial deposit (t): {int(t)} years\nCurrent value of the deposit (V): ${V:.2f}")
            
                
    

# Run the program
if __name__ == "__main__":
    main()    
        
        
        
```

```python
'''
5. Medical Test Results Parser
Create a program that:

- Reads a string containing blood test results (e.g., "Hb: 13.5, WBC: 6.2, Platelets: 250").
- Parses the string into a dictionary of test names and values.
- Handles exceptions for:
- Malformed input (e.g., missing or extra commas).
- Non-numeric values where numeric values are expected.

'''
# Print welcome message to user
print('Welcome to the Medical Test Results Parser')

# Get blood test results from user
results_input = input('Enter a string containing blood test results in this format (e.g., "Hb: 13.5, WBC: 6.2, Platelets: 250"): ')

# Separate individual test results based on the commas.
parsed_results = results_input.split(',')

# Initialize a dictionary to store parsed results
result_dictionary = {}

#Iterate over each test results
for result in parsed_results:
    parsed_blood_result = result.split(':')

    # Validate if each test results consist of two parts
    if len(parsed_blood_result) == 2:
        test_name = parsed_blood_result[0].strip()
        test_value = parsed_blood_result[1].strip()

        # Validate if values before adding to dictionary
        try:
            result_dictionary[test_name] = float(test_value)
        except ValueError:
            print(f'Invalid value for {test_name} : {test_value}')

    # Handle cases user input is malformed (e.g., missing or extra commas)
    else:
        print(f'Malformed input {result}.')
            
# Display parsed  test results 
print(f'\nParsed Blood test results: {result_dictionary}')     



```

```python
'''
Word Counter
Objective: Write a program to count the number of words in a text file.

Steps:
Read the content of the file.
Split the content into words.
Count the total words and print the result.

Bonus: 
Count the occurrences of each unique word.
'''
# Define the path to the file
file_path = "C:/Users/DELL/Downloads/Associate data scientist track.txt"

# Initialize a dictionary to store the occurrences of each word
word_counts = {}

# Open the file in read mode ('r') using a context manager
with open(file_path, 'r') as file:
    # Read all lines from the file
    lines = file.readlines()

    # Loop through each line in the file
    for line in lines:
        # Remove leading/trailing whitespaces and newline characters
        line = line.strip()
        
        # Skip empty lines
        if not line:
            continue
        
        # Split the line into words (default splits by whitespace)
        words = line.split()
        
        # Count each word in the line
        for word in words:
            # Normalize words to lowercase to ensure case-insensitive counting
            word = word.lower()
            
            # Remove punctuation from the word
            word = ''.join(char for char in word if char.isalnum())
            
            # Update word count in the dictionary
            if word in word_counts:
                word_counts[word] += 1
            else:
                word_counts[word] = 1

# Print the total word count
total_words = sum(word_counts.values())
print(f"Total words in the file: {total_words}")

# Print the occurrences of each unique word
print("\nOccurrences of each unique word:")
for word, count in word_counts.items():
    print(f"{word}: {count}")
```

```python
# Define the path to the file where notes will be saved
file_path = "personal_notes.txt"

# Function to write notes to the file
def write_notes():
    # Prompt the user to input text for the note
    note = input("Enter your personal note (or 'exit' to stop): ")

    # Open the file in append mode to add the note without overwriting previous ones
    with open(file_path, 'a') as file:
        file.write(note + '\n')  # Write the note and add a newline

    print("Note saved successfully.")

# Function to read and display the notes from the file
def read_notes():
    # Check if the file exists and read the content
    try:
        with open(file_path, 'r') as file:
            content = file.read()  # Read the entire content of the file
        print("\nYour Personal Notes:")
        print(content)  # Display the notes
    except FileNotFoundError:
        print("No notes have been saved yet.")

# Main loop to give options to the user
while True:
    print("\nPersonal Notes Application")
    print("1. Write a new note")
    print("2. Read notes")
    print("3. Exit")

    choice = input("Please select an option (1/2/3): ")

    if choice == '1':
        write_notes()
    elif choice == '2':
        read_notes()
    elif choice == '3':
        print("Goodbye!")
        break  # Exit the loop and end the program
    else:
        print("Invalid choice. Please try again.")
```

```python
import datetime

# Define the path to the log file
log_file_path = "logfile.txt"

# Function to write a log entry with a timestamp
def write_log():
    # Get the current timestamp
    timestamp = datetime.datetime.now().strftime('%Y-%m-%d %H:%M:%S')
    
    # Prompt the user for a log message
    log_message = input("Enter your log message (or 'exit' to stop): ")

    # Check if the user wants to exit
    if log_message.lower() == 'exit':
        return False  # End the logging loop

    # Open the log file in append mode ('a') to add new log entries without overwriting
    with open(log_file_path, 'a') as log_file:
        log_file.write(f"{timestamp} - {log_message}\n")  # Write timestamp and message

    print(f"Log entry saved at {timestamp}.")
    return True  # Continue logging

# Main loop to collect and save log entries
while True:
    print("\nLog Entry Application")
    print("Enter log messages to be saved with a timestamp.")
    print("Type 'exit' to stop.")
    
    if not write_log():
        print("Exiting the log entry application.")
        break  # Exit the program
```

```python
import string

# Define the path to the file
file_path = "C:/Users/DELL/Downloads/Associate data scientist track.txt"

# Function to count word frequencies
def word_frequency_analysis():
    # Initialize an empty dictionary to store word counts
    word_counts = {}

    # Open the file in read mode
    with open(file_path, 'r') as file:
        # Read all lines from the file
        content = file.read()

        # Remove punctuation from the content
        content = content.translate(str.maketrans('', '', string.punctuation))

        # Split the content into words
        words = content.split()

        # Count each word in the list
        for word in words:
            word = word.lower()  # Convert word to lowercase for case-insensitive counting
            if word in word_counts:
                word_counts[word] += 1
            else:
                word_counts[word] = 1

    # Sort the words by frequency in descending order
    sorted_word_counts = sorted(word_counts.items(), key=lambda x: x[1], reverse=True)

    # Print the sorted word frequencies
    print("Word Frequency Analysis:")
    for word, count in sorted_word_counts:
        print(f"{word}: {count}")

# Run the word frequency analysis
word_frequency_analysis()
```
